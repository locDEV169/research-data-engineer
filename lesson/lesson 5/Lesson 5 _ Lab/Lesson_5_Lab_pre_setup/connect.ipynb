```python
from google.oauth2 import service_account
from google.cloud import bigquery
import google.auth

credentials, adc_project = google.auth.default()
client = bigquery.Client()

df = client.query("""
    SELECT SESSION_USER() AS authenticated_account
""").to_dataframe()

path = ""

# đọc file private key từ gg
creds = service_account.Credentials.from_service_account_file(path)

client = bigquery.Client(credentials=creds, project=creds.project_id)

print("Tài khoản đang kết nối:", df.loc[0, "authenticated_account"])
print("Project của ADC:", adc_project)
print("Project BigQuery client:", client.project)
print("Loại credential:", type(credentials).__name__)
```
